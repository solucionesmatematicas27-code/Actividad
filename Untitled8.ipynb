```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
geometria_polinomios.py
Herramientas para aplicar polinomios a la geometría (K[x], curvas algebraicas,
cuerpos finitos y curvas de Bézier).

Requiere: sympy, numpy, matplotlib      (pip install sympy numpy matplotlib)

Uso rápido (sin argumentos, o en Spyder con F5, se ejecuta la demo):
    python geometria_polinomios.py demo
    python geometria_polinomios.py euclides "x**3+2*x+1" "x**2+1"
    python geometria_polinomios.py factorizar "x**4-x**2-3"
    python geometria_polinomios.py conica "x**2+y**2-4"
    python geometria_polinomios.py interseccion "x**2+y**2-4" "y-x**2+1"
    python geometria_polinomios.py singulares "y**2-x**3-x**2"
    python geometria_polinomios.py eliptica 2 3 97
    python geometria_polinomios.py proyectiva "x**2-y**2-1"
    python geometria_polinomios.py proyectiva "x+y-1" "x+y-3"
    python geometria_polinomios.py animar -1 1 -0.8 1.5     (a b xP xQ -> GIF de P+Q)
    python geometria_polinomios.py bezier "0,0;1,3;3,3;4,0"
    python geometria_polinomios.py interpolar "0,1;1,3;2,2;3,5"
Los gráficos se guardan en la carpeta ./salida
En Spyder, tras ejecutar el archivo puedes llamar las funciones directamente:
    interseccion(x**2 + y**2 - 4, y - x**2 + 1)
    mostrar_proyectiva(x**2 - y**2 - 1)
"""
import argparse
import builtins
import math
import os

import matplotlib
# En Spyder/IPython se muestran los gráficos; en terminal solo se guardan los PNG.
INTERACTIVO = hasattr(builtins, "__IPYTHON__")
if not INTERACTIVO:
    matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np
import sympy as sp

x, y = sp.symbols("x y")
CARPETA = "salida"
COLORES = ["#22693C", "#961E1E", "#1B2A6B", "#C86414", "#5A5A5A"]


def leer(texto):
    """Convierte texto (admite ^ como potencia) en expresión de sympy."""
    return sp.sympify(texto.replace("^", "**"), locals={"x": x, "y": y})


def guardar(fig, nombre):
    os.makedirs(CARPETA, exist_ok=True)
    ruta = os.path.join(CARPETA, nombre)
    fig.savefig(ruta, dpi=150, bbox_inches="tight")
    if INTERACTIVO:
        plt.show()
    else:
        plt.close(fig)
    print(f"  -> gráfico guardado en {ruta}")


# =====================================================================
# 1. POLINOMIOS EN K[x]: división, Euclides, Bézout, factorización
# =====================================================================
def euclides(p, q):
    """Algoritmo de Euclides en Q[x]. Devuelve los pasos y el mcd mónico."""
    a, b = sp.Poly(p, x, domain="QQ"), sp.Poly(q, x, domain="QQ")
    pasos = []
    while not b.is_zero:
        c, r = sp.div(a, b)
        pasos.append((a.as_expr(), b.as_expr(), c.as_expr(), r.as_expr()))
        a, b = b, r
    return pasos, a.monic().as_expr()


def mostrar_euclides(p, q):
    pasos, d = euclides(p, q)
    print(f"\nAlgoritmo de Euclides para p = {p},  q = {q}")
    for i, (a, b, c, r) in enumerate(pasos, 1):
        print(f"  {i}) ({a}) = ({b})·({c}) + ({r})")
    print(f"  mcd(p, q) = {d}")
    s, t, h = sp.gcdex(p, q, x)           # s·p + t·q = h
    ok = sp.expand(s * p + t * q - h) == 0
    print(f"  Bézout: ({sp.expand(s)})·p + ({sp.expand(t)})·q = {h}   [verificado: {ok}]")
    print("  ¿Coprimos?", "sí" if sp.degree(h, x) == 0 else "no")


def eisenstein(p):
    """Devuelve un primo que cumple Eisenstein para p (o None)."""
    c = sp.Poly(p, x).all_coeffs()        # coeficiente líder primero
    if len(c) < 2 or any(not ci.is_integer for ci in c):
        return None
    c = [int(ci) for ci in c]
    for q in sp.primerange(2, max(abs(ci) for ci in c) + 2):
        if c[0] % q and all(ci % q == 0 for ci in c[1:]) and c[-1] % (q * q):
            return q
    return None


def factorizar_reales(p):
    """Factorización en R[x]: factores lineales y cuadráticos irreducibles."""
    out = []
    for f, m in sp.sqf_list(p, x)[1]:
        for r in sp.Poly(f, x).nroots(n=15, maxsteps=200):
            re, im = float(sp.re(r)), float(sp.im(r))
            if abs(im) < 1e-9:
                out.append((f"(x {'-' if re >= 0 else '+'} {abs(re):.6g})", m))
            elif im > 0:
                lin = "" if abs(re) < 1e-9 else f" {'-' if re > 0 else '+'} {abs(2*re):.6g}x"
                out.append((f"(x²{lin} + {re*re+im*im:.6g})", m))
    return out


def mostrar_factorizacion(p):
    print(f"\nFactorización de p = {p}")
    print("  En Q[x]:", sp.factor(p))
    print("  En R[x]:", " · ".join(f"{f}^{m}" if m > 1 else f for f, m in factorizar_reales(p)))
    print("  Raíces en C (con multiplicidad de cada factor libre de cuadrados):")
    for f, m in sp.sqf_list(p, x)[1]:
        for r in sp.Poly(f, x).nroots(n=8, maxsteps=200):
            print(f"     {sp.N(r, 6)}   multiplicidad {m}")
    q = eisenstein(p)
    print("  Eisenstein:", f"se cumple con p = {q} ⇒ irreducible en Q[x]" if q else "no aplica (no concluye nada)")
    g = sp.gcd(p, sp.diff(p, x))
    print("  mcd(p, p') =", g, "→", "raíces simples" if sp.degree(g, x) <= 0 else "hay raíces múltiples")


def interpolar(puntos):
    """Polinomio de menor grado que pasa por los puntos dados."""
    P = sp.expand(sp.interpolate([(sp.nsimplify(u), sp.nsimplify(v)) for u, v in puntos], x))
    print(f"\nPolinomio interpolante (grado {sp.degree(P, x)}): {P}")
    xs = [a for a, _ in puntos]
    t = np.linspace(min(xs) - 0.5, max(xs) + 0.5, 400)
    f = sp.lambdify(x, P, "numpy")
    fig, ax = plt.subplots(figsize=(6, 4))
    ax.plot(t, f(t) * np.ones_like(t), color=COLORES[0])
    ax.scatter(*zip(*puntos), color=COLORES[1], zorder=3)
    ax.grid(alpha=.3)
    ax.set_title("Interpolación polinómica")
    guardar(fig, "interpolacion.png")


# =====================================================================
# 2. CURVAS ALGEBRAICAS EN EL PLANO: cónicas, intersección, singulares
# =====================================================================
def clasificar_conica(f):
    P = sp.Poly(f, x, y)
    if P.total_degree() != 2:
        print("  (no es una cónica: el grado total no es 2)")
        return
    g = lambda i, j: P.coeff_monomial(x**i * y**j)
    A, B, C, D, E, F = g(2, 0), g(1, 1), g(0, 2), g(1, 0), g(0, 1), g(0, 0)
    M = sp.Matrix([[A, B/2, D/2], [B/2, C, E/2], [D/2, E/2, F]])
    disc = B**2 - 4*A*C
    print(f"\nCónica: {sp.expand(f)} = 0")
    print(f"  Discriminante B²-4AC = {disc};  det(matriz 3x3) = {M.det()}")
    if M.det() == 0:
        print("  Tipo: DEGENERADA (rectas, un punto o vacía)")
    elif disc < 0:
        print("  Tipo: elipse" + (" (circunferencia)" if A == C and B == 0 else ""))
    elif disc == 0:
        print("  Tipo: parábola")
    else:
        print("  Tipo: hipérbola")


def interseccion(f, g):
    """Puntos comunes de dos curvas; compara con la cota de Bézout."""
    sols = sp.solve([f, g], [x, y], dict=True)
    reales = []
    for s in sols:
        vx, vy = complex(sp.N(s[x])), complex(sp.N(s[y]))
        if abs(vx.imag) < 1e-9 and abs(vy.imag) < 1e-9:
            reales.append((vx.real, vy.real))
    m, n = sp.Poly(f, x, y).total_degree(), sp.Poly(g, x, y).total_degree()
    print(f"\nIntersección de  f = {f}  y  g = {g}")
    print(f"  Soluciones en C (sin multiplicidad): {len(sols)}   |   cota de Bézout m·n = {m}·{n} = {m*n}")
    print(f"  Soluciones reales: {len(reales)}")
    for px, py in reales:
        print(f"     ({px:.5f}, {py:.5f})")
    res = sp.factor(sp.resultant(f, g, y))
    print("  Resultante respecto a y (polinomio en x que contiene las abscisas):", res)
    graficar([f, g], "interseccion.png", puntos=reales)
    return reales


def singulares(f):
    """Puntos singulares: f = fx = fy = 0 (ahí la curva no es suave)."""
    sols = sp.solve([f, sp.diff(f, x), sp.diff(f, y)], [x, y], dict=True)
    print(f"\nPuntos singulares de {f} = 0: ", sols if sols else "ninguno (curva suave)")
    pts = [(float(s[x]), float(s[y])) for s in sols if s[x].is_real and s[y].is_real]
    graficar([f], "singulares.png", puntos=pts)


def tangente(f, P):
    fx, fy = sp.diff(f, x), sp.diff(f, y)
    gx, gy = fx.subs({x: P[0], y: P[1]}), fy.subs({x: P[0], y: P[1]})
    return sp.Eq(gx * (x - P[0]) + gy * (y - P[1]), 0)


def graficar(curvas, nombre, rango=5, puntos=()):
    xs = np.linspace(-rango, rango, 800)
    X, Y = np.meshgrid(xs, xs)
    fig, ax = plt.subplots(figsize=(6, 6))
    leyenda = []
    for i, f in enumerate(curvas):
        Z = sp.lambdify((x, y), f, "numpy")(X, Y) * np.ones_like(X)
        ax.contour(X, Y, Z, levels=[0], colors=[COLORES[i % 5]], linewidths=2)
        leyenda.append(Line2D([0], [0], color=COLORES[i % 5], lw=2, label=f"{sp.expand(f)} = 0"))
    if len(puntos):
        ax.scatter(*zip(*puntos), color="black", zorder=4, s=35)
    ax.axhline(0, color="gray", lw=.5); ax.axvline(0, color="gray", lw=.5)
    ax.set_aspect("equal"); ax.grid(alpha=.25); ax.legend(handles=leyenda, loc="upper right", fontsize=8)
    guardar(fig, nombre)


# =====================================================================
# 3. GEOMETRÍA SOBRE CUERPOS FINITOS: curvas elípticas en F_p
# =====================================================================
def suma_puntos(P, Q, a, p):
    """Ley de grupo (cuerda y tangente). None representa el punto en el infinito."""
    if P is None: return Q
    if Q is None: return P
    x1, y1 = P; x2, y2 = Q
    if x1 == x2 and (y1 + y2) % p == 0:
        return None
    if P == Q:
        m = (3*x1*x1 + a) * pow(2*y1, -1, p) % p
    else:
        m = (y2 - y1) * pow(x2 - x1, -1, p) % p
    x3 = (m*m - x1 - x2) % p
    return (x3, (m*(x1 - x3) - y1) % p)


def orden_punto(P, a, p):
    n, R = 1, P
    while True:
        R = suma_puntos(R, P, a, p); n += 1
        if R is None:
            return n


def curva_eliptica(a, b, p):
    if (4*a**3 + 27*b**2) % p == 0:
        raise ValueError("Curva singular módulo p (discriminante 0): elige otros a, b, p.")
    pts = [(i, j) for i in range(p) for j in range(p) if (j*j - (i**3 + a*i + b)) % p == 0]
    print(f"\nCurva elíptica y² = x³ + ({a})x + ({b}) sobre F_{p}")
    print(f"  Número de puntos afines: {len(pts)}  (con el infinito: {len(pts)+1})")
    cota = 2 * math.sqrt(p)
    print(f"  Cota de Hasse: |#E - (p+1)| = {abs(len(pts)+1-(p+1))} ≤ 2√p = {cota:.2f}")
    if pts:
        P = pts[0]
        print(f"  Punto {P}: orden {orden_punto(P, a, p)};  2P = {suma_puntos(P, P, a, p)}")
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 5))
    xs = np.linspace(-4, 6, 800); X, Y = np.meshgrid(xs, np.linspace(-8, 8, 800))
    ax1.contour(X, Y, Y**2 - (X**3 + a*X + b), levels=[0], colors=[COLORES[0]], linewidths=2)
    ax1.set_title(f"Sobre R: y² = x³ + ({a})x + ({b})"); ax1.grid(alpha=.3)
    ax2.scatter(*zip(*pts), color=COLORES[1], s=25) if pts else None
    ax2.set_xlim(-1, p); ax2.set_ylim(-1, p); ax2.set_aspect("equal"); ax2.grid(alpha=.3)
    ax2.set_title(f"Sobre F_{p}: {len(pts)} puntos afines")
    guardar(fig, "eliptica.png")


# =====================================================================
# 4. CURVAS DE BÉZIER (polinomios de Bernstein)
# =====================================================================
def bezier(ctrl, n=300):
    P = np.array(ctrl, dtype=float)
    g = len(P) - 1
    t = np.linspace(0, 1, n)
    B = np.array([math.comb(g, i) * t**i * (1 - t)**(g - i) for i in range(g + 1)])
    return B.T @ P


def mostrar_bezier(ctrl):
    g = len(ctrl) - 1
    T = sp.symbols("t")
    bx = sp.expand(sum(math.comb(g, i) * T**i * (1 - T)**(g - i) * sp.nsimplify(c[0]) for i, c in enumerate(ctrl)))
    by = sp.expand(sum(math.comb(g, i) * T**i * (1 - T)**(g - i) * sp.nsimplify(c[1]) for i, c in enumerate(ctrl)))
    print(f"\nCurva de Bézier de grado {g}")
    print(f"  x(t) = {bx}\n  y(t) = {by}")
    C = bezier(ctrl); P = np.array(ctrl, dtype=float)
    fig, ax = plt.subplots(figsize=(6, 4.5))
    ax.plot(P[:, 0], P[:, 1], "--o", color=COLORES[4], label="polígono de control")
    ax.plot(C[:, 0], C[:, 1], color=COLORES[0], lw=2.5, label="curva de Bézier")
    ax.legend(); ax.grid(alpha=.3); ax.set_aspect("equal")
    guardar(fig, "bezier.png")


# =====================================================================
# 5. ANIMACIÓN DE LA LEY DE GRUPO EN UNA CURVA ELÍPTICA (sobre R)
# =====================================================================
def animar_eliptica(a, b, xp, xq, nombre="ley_de_grupo.gif"):
    """GIF de P+Q: cuerda (o tangente), tercer punto y reflexión vertical."""
    from matplotlib.animation import FuncAnimation, PillowWriter
    f = lambda t: t**3 + a*t + b
    if f(xp) <= 0 or f(xq) <= 0:
        raise ValueError("xP y xQ deben estar sobre la parte real de la curva (x³+ax+b > 0).")
    if 4*a**3 + 27*b**2 == 0:
        raise ValueError("Curva singular (discriminante 0).")
    P = (xp, math.sqrt(f(xp)))
    Q = (xq, -math.sqrt(f(xq)) if xq != xp else P[1])      # Q abajo; si xq=xp, Q=P (tangente)
    if xp == xq:
        m = (3*xp**2 + a) / (2*P[1])
    elif xp != xq:
        m = (Q[1] - P[1]) / (Q[0] - P[0])
    x3 = m*m - P[0] - Q[0]
    y3 = P[1] + m*(x3 - P[0])                               # tercer punto de corte
    R = (x3, -y3)                                           # P + Q
    print(f"\nLey de grupo sobre R: P = ({P[0]:.3f}, {P[1]:.3f}),  Q = ({Q[0]:.3f}, {Q[1]:.3f})")
    print(f"  pendiente = {m:.4f};  tercer punto = ({x3:.3f}, {y3:.3f});  P+Q = ({R[0]:.3f}, {R[1]:.3f})")

    ext = max(abs(P[0]), abs(Q[0]), abs(x3)) + 1.5
    lo, hi = min(-2.5, P[0] - 1, Q[0] - 1, x3 - 1), ext
    X, Yg = np.meshgrid(np.linspace(lo, hi, 800), np.linspace(-ext - 2, ext + 2, 800))
    fig, ax = plt.subplots(figsize=(6, 6))
    xl = np.array([lo, hi])

    def cuadro(k):
        ax.clear()
        ax.contour(X, Yg, Yg**2 - (X**3 + a*X + b), levels=[0], colors=[COLORES[0]], linewidths=2)
        ax.set_xlim(lo, hi); ax.set_ylim(-ext - 2, ext + 2); ax.grid(alpha=.3); ax.set_aspect("equal")
        ax.scatter(*P, color=COLORES[2], zorder=5, s=50); ax.annotate("P", P, textcoords="offset points", xytext=(6, 6))
        if xp != xq:
            ax.scatter(*Q, color=COLORES[2], zorder=5, s=50); ax.annotate("Q", Q, textcoords="offset points", xytext=(6, 6))
        titulo = "1) Puntos P y Q"
        if k >= 10:                                          # la recta crece desde P
            u = min((k - 10) / 14, 1)
            ax.plot([P[0] + (lo - P[0])*u, P[0] + (hi - P[0])*u],
                    [P[1] + m*(lo - P[0])*u, P[1] + m*(hi - P[0])*u], color=COLORES[3], lw=2)
            titulo = "2) Recta por P y Q" + (" (tangente)" if xp == xq else "")
        if k >= 25:
            ax.scatter(x3, y3, color=COLORES[3], zorder=5, s=50); ax.annotate("tercer punto", (x3, y3), textcoords="offset points", xytext=(6, 6))
            titulo = "3) Tercer punto de corte"
        if k >= 30:
            u = min((k - 30) / 8, 1)
            ax.plot([x3, x3], [y3, y3 + (R[1] - y3)*u], "--", color=COLORES[1], lw=2)
            if u >= 1:
                ax.scatter(*R, color=COLORES[1], zorder=6, s=70); ax.annotate("P + Q", R, textcoords="offset points", xytext=(6, 6))
            titulo = "4) Reflexión respecto al eje x: P + Q"
        ax.set_title(f"y² = x³ + ({a})x + ({b})   —   {titulo}")

    anim = FuncAnimation(fig, cuadro, frames=range(0, 44), interval=120)
    os.makedirs(CARPETA, exist_ok=True)
    ruta = os.path.join(CARPETA, nombre)
    anim.save(ruta, writer=PillowWriter(fps=8))
    plt.close(fig)
    print(f"  -> animación guardada en {ruta}")


# =====================================================================
# 6. CURVAS PROYECTIVAS: homogeneización y puntos en el infinito
# =====================================================================
X_, Y_, Z_ = sp.symbols("X Y Z")


def homogeneizar(f):
    d = sp.Poly(f, x, y).total_degree()
    return sp.expand(Z_**d * f.subs({x: X_/Z_, y: Y_/Z_}, simultaneous=True)), d


def puntos_infinito(f):
    """Puntos [X:Y:0] de la curva, con multiplicidad (en C)."""
    F, d = homogeneizar(f)
    F0 = sp.expand(F.subs(Z_, 0))                           # parte de mayor grado de f
    pts = {}
    g = sp.Poly(F0.subs(Y_, 1), X_)
    if not g.is_zero:
        for r, m in sp.roots(g, multiple=False).items():
            pts[f"[{sp.simplify(r)} : 1 : 0]"] = m
        if d - g.degree() > 0:
            pts["[1 : 0 : 0]"] = d - g.degree()
    return F, d, pts


def mostrar_proyectiva(f, g=None):
    F, d, pts = puntos_infinito(f)
    print(f"\nCurva afín f = {f}  (grado {d})")
    print(f"  Homogeneización F(X,Y,Z) = {F}")
    if pts:
        print("  Puntos en el infinito (Z = 0):", ", ".join(f"{p} (mult. {m})" for p, m in pts.items()))
    else:
        print("  Sin puntos en el infinito.")
    tipo = ""
    if d == 2:
        tipo = {0: "elipse (2 puntos complejos en el infinito)",
                1: "parábola (1 punto en el infinito: tangente a la recta del infinito)",
                2: "hipérbola (2 puntos reales en el infinito: las direcciones de sus asíntotas)"}.get(len(pts), "")
        if len(pts) == 0 or (len(pts) == 2 and any("I" in p for p in pts)):
            tipo = "elipse (2 puntos complejos en el infinito)"
        if tipo:
            print("  Interpretación geométrica:", tipo)
    if g is not None:
        G, e, pg = puntos_infinito(g)
        comunes = [(p, min(pts[p], pg[p])) for p in pts if p in pg]
        print(f"\nCurva afín g = {g}  (grado {e});  puntos en el infinito:", ", ".join(pg) or "ninguno")
        print("  Puntos comunes en el infinito:", ", ".join(f"{p} (mult. ≥ {m})" for p, m in comunes) or "ninguno")
        if d == e == 1 and comunes:
            print("  → Rectas paralelas: se cortan en el infinito (en geometría proyectiva toda pareja de rectas se corta).")
        n_afin = len(sp.solve([f, g], [x, y], dict=True))
        print(f"  Puntos afines comunes: {n_afin};  cota de Bézout = {d*e}")


# =====================================================================
# DEMO Y LÍNEA DE COMANDOS
# =====================================================================
def parse_puntos(texto):
    return [tuple(float(v) for v in par.split(",")) for par in texto.split(";")]


def demo():
    print("=== DEMO: polinomios y geometría ===")
    mostrar_euclides(x**3 + 2*x + 1, x**2 + 1)
    mostrar_euclides(x**2 - 1, x**2 - x)
    mostrar_factorizacion(x**4 - x**2 - 3)
    mostrar_factorizacion((x - 1)**2 * (x**2 + 1))
    f, g = x**2 + y**2 - 4, y - x**2 + 1
    clasificar_conica(f); clasificar_conica(g)
    interseccion(f, g)
    h = y**2 - x**3 - x**2
    print("\nTangente a", h, "= 0 en el punto (3, 6):", tangente(h, (3, 6)))
    singulares(h)
    curva_eliptica(2, 3, 97)
    mostrar_bezier([(0, 0), (1, 3), (3, 3), (4, 0)])
    interpolar([(0, 1), (1, 3), (2, 2), (3, 5)])
    mostrar_proyectiva(x**2 - y**2 - 1)
    mostrar_proyectiva(x + y - 1, x + y - 3)
    animar_eliptica(-1, 1, -0.8, 1.5)


def main():
    ap = argparse.ArgumentParser(description="Geometría y polinomios")
    sub = ap.add_subparsers(dest="cmd")   # sin comando => demo (útil en Spyder)
    sub.add_parser("demo")
    for nombre, nargs in [("euclides", 2), ("interseccion", 2), ("factorizar", 1),
                          ("conica", 1), ("singulares", 1)]:
        s = sub.add_parser(nombre)
        for k in range(nargs):
            s.add_argument(f"expr{k+1}", help="expresión en x, y (usa ** o ^ para potencias)")
    e = sub.add_parser("eliptica")
    for nom in ("a", "b", "p"):
        e.add_argument(nom, type=int)
    pr = sub.add_parser("proyectiva")
    pr.add_argument("expr1"); pr.add_argument("expr2", nargs="?", default=None)
    an = sub.add_parser("animar")
    an.add_argument("a", type=float); an.add_argument("b", type=float)
    an.add_argument("xp", type=float); an.add_argument("xq", type=float)
    sub.add_parser("bezier").add_argument("puntos", help='"x0,y0;x1,y1;..."')
    sub.add_parser("interpolar").add_argument("puntos", help='"x0,y0;x1,y1;..."')
    a = ap.parse_args()

    if a.cmd in (None, "demo"): demo()
    elif a.cmd == "euclides": mostrar_euclides(leer(a.expr1), leer(a.expr2))
    elif a.cmd == "factorizar": mostrar_factorizacion(leer(a.expr1))
    elif a.cmd == "conica": clasificar_conica(leer(a.expr1)); graficar([leer(a.expr1)], "conica.png")
    elif a.cmd == "interseccion": interseccion(leer(a.expr1), leer(a.expr2))
    elif a.cmd == "singulares": singulares(leer(a.expr1))
    elif a.cmd == "eliptica": curva_eliptica(a.a, a.b, a.p)
    elif a.cmd == "proyectiva": mostrar_proyectiva(leer(a.expr1), leer(a.expr2) if a.expr2 else None)
    elif a.cmd == "animar": animar_eliptica(a.a, a.b, a.xp, a.xq)
    elif a.cmd == "bezier": mostrar_bezier(parse_puntos(a.puntos))
    elif a.cmd == "interpolar": interpolar(parse_puntos(a.puntos))


if __name__ == "__main__":
    main()
```
